# Predictive Coding

## Symbols

$x_i^l$ : Activity of neuron $i$ in layer $l$

$\mu_i^l$ : Predicted activity of neuron $i$ in layer $l$

$w_{ik}^l$ : Synaptic weight from neuron $k$ of layer $l$ to neuron $i$ of layer
$l + 1$

$b_i^l$ : Activity bias of neuron $i$ in layer $l$

## Equations

### Prediction

Definition from video:
$$
\mu_i^l = \sum_k{w_{ik}^l \sigma(x_k^{l - 1})}
$$

Naïve addition of bias: (Type 1)
$$
\mu_i^l = \sum_k{w_{ik}^l \sigma(x_k^{l - 1} + b_k^{l - 1})}
$$

Definition closer to traditional neural network: (Type 2)
$$
\mu_i^l = \sigma\left(\sum_k{w_{ik}^l x_k^{l - 1}} + b_i^l\right)
$$

### Energy

$$
\epsilon_i^l = x_i^l - \mu_i^l
$$

$$
E = \sum_{li}{(\epsilon_i^l)^2}
$$

### Neural relaxation

$$
\frac{\partial E}{\partial x_i^l}
=
\sum_{kj}{2 \epsilon_j^k \frac{\partial \epsilon_j^k}{\partial x_i^l}}
$$
$$
\frac{\partial \epsilon_j^k}{\partial x_i^l}
=
\delta_{li,kj} - \frac{\partial \mu_j^k}{\partial x_i^l}
=
\delta_{li,kj} - \delta_{k,l + 1} \frac{\partial \mu_j^{l + 1}}{\partial x_i^l}
$$
$$
\frac{\partial \mu_j^k}{\partial x_i^l}
=
\delta_{k,l + 1} \frac{\partial \mu_j^{l + 1}}{\partial x_i^l}
$$

$$
\frac{\partial E}{\partial x_i^l}
=
\sum_{kj}{2\epsilon_j^k \left(\delta_{li,kj} - \delta_{k,l + 1} \frac{\partial \mu_j^{l + 1}}{\partial x_i^l}\right)}
=
2 \epsilon_i^l
-
2 \sum_j{\epsilon_j^{l + 1} \frac{\partial \mu_j^{l + 1}}{\partial x_i^l}}
$$

#### Type 1:

$$
\mu_j^{l + 1}
=
\sum_k{w_{jk}^{l + 1} \sigma\left(x_k^l + b_k^l\right)}
$$
$$
\frac{\partial \mu_j^{l + 1}}{\partial x_i^l}
=
w_{ji}^{l + 1} \sigma'\left(x_i^l + b_i^l\right)
$$

$$
\frac{\partial E}{\partial x_i^l}
=
2 \epsilon_i^l
-
2 \sum_j{\epsilon_j^{l + 1} w_{ji}^{l + 1} \sigma'\left(x_i^l + b_i^l\right)}
$$

#### Type 2:

$$
\mu_j^{l + 1} = \sigma(z_j^{l + 1})
$$
$$
z_j^{l + 1} = \sum_k{w_{jk}^{l + 1} x_k^l} + b_j^{l + 1}
$$

$$
\frac{\partial \mu_j^{l + 1}}{\partial x_i^l}
=
\sigma'(z_j^{l + 1}) \frac{\partial z_j^{l + 1}}{\partial x_i^l}
=
\sigma'(z_j^{l + 1}) w_{ji}^{l + 1}
$$

$$
\frac{\partial E}{\partial x_i^l}
=
2 \epsilon_i^l
-
2 \sum_j{\epsilon_j^{l + 1} \sigma'(z_j^{l + 1}) w_{ji}^{l + 1}}
$$
